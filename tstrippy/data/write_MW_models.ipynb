# Milky Way Models

```python
import tstrippy
from pathlib import Path
from astropy import units as u 
from astropy import constants as const
import yaml
import numpy as np

current_script_directory = Path(".").parent
path_to_data = current_script_directory / ".." / "data" 
path_to_MWmodels = path_to_data / "MWmodels"
absolute_path_to_MWmodels = path_to_MWmodels.resolve()
print(absolute_path_to_MWmodels)
```

## Pouliasis 2017
[Pouliasis 2017 et al.](10.1051/0004-6361/201527346) 

In their unit basis, G=1, V=1, L=1 
They also use V = 10 km/s (which is implied since the potential is given in units of 100 km2/s2)
when putting in the standard gravitational constnat, km/s, kpc, the unit of mass becomes: 

```python
## model 1 
unitL = u.kpc 
unitV = 10 * u.km / u.s
unitM = (u.kpc* (unitV**2) /const.G).to(u.Msun)
display(unitM)
```

### Model 1 

from table 1 in their paper 

```python
Mhalo = 6000 * unitM 
ahalo = 14 * unitL
rcut = 100 * unitL
gamma = 2.02
component0 = {
    "name": "allensantillianhalo",
    "parameters": [Mhalo.to(u.Msun).value, ahalo.value, rcut.value, gamma],
    "parameter_names": ["M", "a", "rcut", "gamma"],
    "parameter_units": [Mhalo.to(u.Msun).unit, ahalo.unit, rcut.unit, ""]
}

Mthin = 1700 * unitM
athin = 5.3 * unitL
bthin = 0.25 * unitL
component1 = {
    "name": "miyamotonagai",
    "parameters": [Mthin.to(u.Msun).value, athin.value, bthin.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mthin.to(u.Msun).unit, athin.unit, bthin.unit, ]
}

name = "miyamotonagai"
Mthick = 1700 * unitM 
athick = 2.6 * unitL
bthick = 0.8 * unitL
component2 = {
    "name": "miyamotonagai",
    "parameters": [Mthick.to(u.Msun).value, athick.value, bthick.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mthick.to(u.Msun).unit, athick.unit, bthick.unit, ]
}

name = "plummer"
Mbuldge = 460 * unitM
bbulge = 0.3 * unitL 
component3 = {
    "name": "plummer",
    "parameters": [Mbuldge.to(u.Msun).value, bbulge.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mbuldge.to(u.Msun).unit, bbulge.unit, ]
}

model = {"name": "pouliasis2017p1", "components": [component0, component1, component2, component3]}
model = tstrippy.io.milkyway_models._to_yaml_compatible(model)

# a user would do: tstrippy.io.milkyway_models.save_model_yaml() for their personal models
# But here, we're creating the standards in the package
fname = absolute_path_to_MWmodels.as_posix() + "/" +model['name'] +".yaml"
with open(fname, "w") as fp:
    yaml.safe_dump(model,fp,sort_keys=False)
```

## Ibata 2024

[the paper](10.3847/1538-4357/ad382d)

These values are mostly in table 2, but you have to also refer to [McMilian 2017](10.1093/mnras/stw2759) for the gaseous disks

```python
# max beta free
ibata2024halo = {
    "name": "ibata2024halo",
    "parameters": [11200000, 19.5, 1000, 0.727, 1.14, 2.54],
    "parameter_names": ["rho0", "r0", "rt", "q", "gamma", "beta"],
    "parameter_units": ["solMass / kpc3", "kpc", "kpc", "", "", ""]}

Md = 3.65e10
Rd = 2.251 ; zd = 0.349
simgaD = float(Md / (2*np.pi*Rd**2))
exponentialdisk1 = {
    "name": "exponentialdisk",
    "parameters": [simgaD, Rd, zd],
    "parameter_names": ["Sigma0", "hR", "hz"],
    "parameter_units": ["solMass / kpc2", "kpc", "kpc"]}

Mt = 0.62e10
Rt = 2.07 ; zt = 0.860
simgat = float(Mt / (2*np.pi*Rd**2))
exponentialdisk2 = {
    "name": "exponentialdisk",
    "parameters": [simgat, Rt, zt],
    "parameter_names": ["Sigma0", "hR", "hz"],
    "parameter_units": ["solMass / kpc2", "kpc", "kpc"]}

simga = 52.1 * u.Msun/(u.pc)**2
simga=float(simga.to(u.Msun/u.kpc**2).value)
R = 7 ; z = 0.085 ; Rm = 4
h1disc =  {
    "name": "mcmilan2017disk",
    "parameters": [simga, R, z, Rm],
    "parameter_names": ["Sigma0", "hR", "hz", "Rm"],
    "parameter_units": ["solMass / kpc2", "kpc", "kpc"]}

simga = 2180 * u.Msun/(u.pc)**2
simga=float(simga.to(u.Msun/u.kpc**2).value)
R = 1.5 ; z = 0.045 ; Rm = 12
h2disc =  {
    "name": "mcmilan2017disk",
    "parameters": [simga, R, z, Rm],
    "parameter_names": ["Sigma0", "hR", "hz", "Rm"],
    "parameter_units": ["solMass / kpc2", "kpc", "kpc"]}

components = [ibata2024halo, exponentialdisk1, exponentialdisk2, h1disc, h2disc]
model = {"name": "ibata2024mostprob", "components": components}
model = tstrippy.io.milkyway_models._to_yaml_compatible(model)
fname = absolute_path_to_MWmodels.as_posix() + "/" +model['name'] +".yaml"
with open(fname, "w") as fp:
    yaml.safe_dump(model,fp,sort_keys=False)
```
