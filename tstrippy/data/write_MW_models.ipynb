# Milky Way Models

```python
import tstrippy
from pathlib import Path
from astropy import units as u 
from astropy import constants as const
import yaml

current_script_directory = Path(".").parent
path_to_data = current_script_directory / ".." / "data" 
path_to_MWmodels = path_to_data / "MWmodels"
absolute_path_to_MWmodels = path_to_MWmodels.resolve()
print(absolute_path_to_MWmodels)
```

## Pouliasis 2017
[Pouliasis 2017 et al.](10.1051/0004-6361/201527346) 

In their unit basis, G=1, V=1, L=1
when putting in the standard gravitational constnat, km/s, kpc, the unit of mass becomes: 

```python
## model 1 
unitL = u.kpc 
unitM = (u.kpc* ((u.km/u.s)**2) /const.G).to(u.Msun)
display(unitM)
```

### Model 1 

from table 1 in their paper 

```python
Mhalo = 6000 * unitM 
ahalo = 14 * unitL
rcut = 100 * unitL
gamma = 2.02
component0 = {
    "name": "allensantillianhalo",
    "parameters": [Mhalo.to(u.Msun).value, ahalo.value, rcut.value, gamma],
    "parameter_names": ["M", "a", "rcut", "gamma"],
    "parameter_units": [Mhalo.to(u.Msun).unit, ahalo.unit, rcut.unit, ""]
}

Mthin = 1700 * unitM
athin = 5.3 * unitL
bthin = 0.25 * unitL
component1 = {
    "name": "miyamotonagai",
    "parameters": [Mthin.to(u.Msun).value, athin.value, bthin.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mthin.to(u.Msun).unit, athin.unit, bthin.unit, ]
}

name = "miyamotonagai"
Mthick = 1700 * unitM 
athick = 2.6 * unitL
bthick = 0.8 * unitL
component2 = {
    "name": "miyamotonagai",
    "parameters": [Mthick.to(u.Msun).value, athick.value, bthick.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mthick.to(u.Msun).unit, athick.unit, bthick.unit, ]
}

name = "plummer"
Mbuldge = 460 * unitM
bbulge = 0.3 * unitL 
component3 = {
    "name": "plummer",
    "parameters": [Mbuldge.to(u.Msun).value, bbulge.value],
    "parameter_names": ["M", "a", "b", ],
    "parameter_units": [Mbuldge.to(u.Msun).unit, bbulge.unit, ]
}

model = {"name": "pouliasis2017pi", "components": [component0, component1, component2, component3]}
model = tstrippy.io.milkyway_models._to_yaml_compatible(model)


# a user would do: tstrippy.io.milkyway_models.save_model_yaml() for their personal models
# But here, we're creating the standards in the package
fname = absolute_path_to_MWmodels.as_posix() + "/" +model['name'] +".yaml"
with open(fname, "w") as fp:
    yaml.safe_dump(model,fp,sort_keys=False)
```
